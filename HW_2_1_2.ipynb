1. Задана функція залежності оцінки від кількості годин, витрачених на навчання: $ f(x) = 10x - 5 $.
  1. З допомогою Python обчисліть похідну цієї функції.
  2. Використовуйте бібліотеку `matplotlib` для побудови графіків функції та її похідної в діапазоні значень від 0 до 10 годин. Проаналізуйте, як зміна кількості годин вивчення впливає на оцінку.



```python
import matplotlib.pyplot as plt
import numpy as np
import sympy as sp
```

```python
x = sp.symbols('x')
f_lin = 10 * x - 5
derivative_f_lin = sp.diff(f_lin, x)
print(f"Похідна від f(x) = 10x - 5 -> {derivative_f_lin}")
```

```python
hours = np.linspace(0, 10, 100)
grade = 10 * hours - 5
df = np.full_like(hours, 10)

plt.figure(figsize=(10, 5))
plt.plot(hours, grade, color='blue')
plt.plot(hours, df, color='red')

plt.title("Вплив кількості годин на оцінку")
plt.xlabel('Години навчання')
plt.ylabel('Швидкість зміни оцінки')


plt.show()
```

Зміна годин впливає на оцінку лінійно: кожна година = +10 балів, незалежно від того, перша вона чи десята.

2. Використовуючи Python і `SymPy`, обчисліть похідні для функцій $ f(x) = \sin(x) $ та $ f(x) = \ln(x) $. Зобразіть ці функції та їхні похідні за допомогою `matplotlib`.



```python
x = sp.symbols('x')
f_sin = sp.sin(x)
f_ln = sp.ln(x)
derivative_f_sin = sp.diff(f_sin, x)
derivative_f_ln = sp.diff(f_ln, x)
print(f"Похідна від f_sin -> {derivative_f_sin}")
print(f"Похідна від f_ln -> {derivative_f_ln}")
```

```python
x = np.linspace(0.1, 10, 100)
f = np.sin(x)
df = np.cos(x)

plt.figure(figsize=(10, 5))
plt.plot(x, f, color='blue')
plt.plot(x, df, color='red')

plt.title("Перша формула")
plt.xlabel('Значення x')
plt.ylabel('Значення y')


plt.show()
```

```python
x = np.linspace(0.1, 10, 100)
f = np.log(x)
df = 1/x

plt.figure(figsize=(10, 5))
plt.plot(x, f, color='blue')
plt.plot(x, df, color='red')

plt.title("Друга формула")
plt.xlabel('Значення x')
plt.ylabel('Значення y')


plt.show()
```


3. Загальні витрати компанії від обсягу виробництва задані функцією $ f(x) = x^2 - 5x + 1 $, де $ x $ — кількість вироблених одиниць. Обчисліть похідну цієї функції з допомогою SymPy та візуалізуйте значення функції і її похідної на відрізку від 0 до 100.

```python
x = sp.symbols('x')

f_x = x ** 2 - 5 * x +1
derivative_f_x = sp.diff(f_x, x)
print("Похідна: ", derivative_f_x)
```

```python
x = np.linspace(0, 100, 100)
f = x ** 2 - 5 * x +1
df = 2*x - 5

plt.figure(figsize=(10, 5))
plt.plot(x, f, color='blue')
plt.plot(x, df, color='red')

plt.title('Витрати компанії та їхня похідна')
plt.xlabel('Обсяг виробництва (одиниць)')
plt.ylabel('Витрати')


plt.show()
```

```python
fig, ax1 = plt.subplots(figsize=(10, 5))

ax1.plot(x, f, color='blue', label='f(x) = x² - 5x + 1')
ax1.set_xlabel('Обсяг виробництва')
ax1.set_ylabel('Витрати', color='blue')

ax2 = ax1.twinx()
ax2.plot(x, df, color='red', label="f'(x) = 2x - 5")
ax2.set_ylabel('Граничні витрати', color='red')

plt.title('Витрати компанії та їхня похідна')
fig.tight_layout()
plt.show()
```
